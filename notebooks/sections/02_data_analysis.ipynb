<h3>2.2.1 Interpretation of CSV File's Data Column</h3>

<p><strong><u>DATE</u></strong></p>
<ul>
    <li>Represents the timestamp for each recorded entry.</li>
    <li>Indicates the period (daily, monthly, or yearly) for energy consumption measurements.</li>
</ul>

<p><strong><u>Gas Consumption (tons)</u></strong></p>
<ul>
    <li>Measures the amount of gas consumed in tons for the given date.</li>
    <li>Helps analyze trends in gas usage over time for efficiency planning.</li>
</ul>

<p><strong><u>Electricity Consumption (MWh)</u></strong></p>
<ul>
    <li>Represents the total electricity usage in megawatt-hours (MWh) for each recorded date.</li>
    <li>Used to monitor power demand and identify seasonal patterns.</li>
</ul>

<p><strong><u>Water Consumption (tons)</u></strong></p>
<ul>
    <li>Measures the water usage in tons for the given date.</li>
    <li>Useful for understanding water consumption trends and optimizing resource allocation.</li>
</ul>
