## Assessment Objective

Train time-series models and forecast gas, electricity, and water consumption for the next 60 months.

## Background

Accurate consumption forecasting helps organisations plan resource allocation, manage cost, and reason about sustainability. This notebook applies ARIMA and SARIMAX models to historical monthly consumption data and compares their suitability for gas, electricity, and water forecasting.
