<p>In this section, I will be conducting Data Cleaning. Throughout this section, I will be checking, imputing and verifying missing datas for my datasets.</p>

<hr/>
<h2>3.1 Checking For Missing Data</h2>

<p>In the cell below, I will be checking for the total amount of missing data value per category. By checking for the sum of the missing data, I can later compare it to the sum of missing data after I imputed the missing data.</p>

```python
# ----------- Placing Missing Values of df Into A Dataframe ----------- #
missingData = pd.DataFrame({'Missing Values': df.isnull().sum().values}, index = df.columns)

# ----------- Function for Colour Coding ----------- #
def colour_code(values):
    return 'background-color: green' if values == 0 else 'background-color: red'

# ----------- Calling Function for Colour Coding ----------- #
missingData = missingData.style.map(colour_code, subset = ['Missing Values'])

# ----------- Displaying Missing Values ----------- #
missingData
```

<p>With reference to the output above, I am able to verify that there is no missing data within the dataset and that there is no need to impute any missing data for any of the Data Column. This allows me to subsequently proceed to check for Duplicated Data within the dataset.</p>

<hr/>
<h2>3.2 Checking For Duplicated Data</h2>

<p>To address Duplicated Data, I have to verify if there are duplicated data in the first place. In the cell below, I will be checking for duplicated data and based on the output, one of the two action plan will occur.</p>
<br/>
<span style="color:red"><p><strong><u>Duplicated Data Identified</u></strong></p></span>
<ul>
    <li>Proceed to remove the duplicated data.</li>
</ul>
<span style="color:green"><p><strong><u>No Duplicated Data Identified</u></strong></p></span>
<ul>
    <li>Proceed with next sub-section on Fixing Data Types.</li>
</ul>
<br/>
<p>With the action plan listed above, I will be proceeding to check for duplicated data.</p>

```python
# ----------- Calculate Total Duplicated Rows ----------- #
duplicatesCount = df.duplicated().sum()

# ----------- Create DataFrame to Store Result ----------- #
duplicateSummary = pd.DataFrame({'Count': [duplicatesCount]}, index = ['Duplicated Rows'])

# ----------- Function for Colour Coding ----------- #
def colour_code(val):
    if val > 0:
        return 'background-color: red; color: white'
    else:
        return 'background-color: green; color: white'

# ----------- Apply Colour Coding to the 'Count' Column ----------- #
duplicateSummary = duplicateSummary.style.map(colour_code, subset=['Count'])

# ----------- Display the Result ----------- #
duplicateSummary
```

<p>With reference to the output of the Dataframe above, I am able to verify that there are <span style="color:green"><strong><u>No Duplicated Data Identified</u></strong></span>. Therefore, I am able to proceed to Feature Engineering for the Dataset.</p>

<h2>3.3 Correcting Data Types</h2>

<p>In this sub-section, I will be correcting the Date Data Column from an object into a DateTime format. The operation will be conducted in the cell below.</p>

```python
# ------------- Convert DATE Data Column To DateTime Format ------------- #
df['DATE'] = pd.to_datetime(df['DATE'], format='%d/%m/%Y')

# ------------- Set DATE As Index ------------- #
df.set_index('DATE', inplace=True)

# ------------- Displaying DataFrame -------------#
df.head(5).style.background_gradient(cmap="Blues")
```

<p>With reference to the output of the DataFrame above, I am able to see that the conversion has been done successfully.</p>
